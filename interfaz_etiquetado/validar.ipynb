```python
import os
import cv2
import pandas as pd

# === CONFIGURACIÓN ===
ROOT_DIR = r"D:/la-u/ciclo 2025-1/Seminario/DATASET_UNILADO"
CSV_PATH = r"D:/la-u/ciclo 2025-1/Seminario/Proyecto/interfaz_etiquetado/etiquetas_por_frame.csv"

# Cargar CSV
df = pd.read_csv(CSV_PATH)
print(f"✅ CSV cargado: {len(df)} filas")

# Obtener lista de videos
def obtener_videos_recursivo(root_dir):
    video_paths = []
    for root, _, files in os.walk(root_dir):
        for file in files:
            if file.lower().endswith((".mp4", ".avi", ".mov", ".mkv")):
                video_paths.append(os.path.join(root, file))
    video_paths.sort()
    return video_paths

videos = obtener_videos_recursivo(ROOT_DIR)

# === Comparar frames existentes vs etiquetados ===
resumen = []

for vpath in videos:
    vname = os.path.basename(vpath)

    # Obtener total de frames reales del video
    cap = cv2.VideoCapture(vpath)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    cap.release()

    # Frames etiquetados
    frames_etiquetados = df[df["video"] == vname]["frame"].nunique()

    completo = frames_etiquetados == total_frames
    faltan = total_frames - frames_etiquetados

    resumen.append({
        "video": vname,
        "frames_totales": total_frames,
        "frames_etiquetados": frames_etiquetados,
        "faltan": faltan,
        "completo": completo
    })

# Mostrar resumen como DataFrame
resumen_df = pd.DataFrame(resumen)
display(resumen_df)

# Mostrar videos incompletos
incompletos = resumen_df[~resumen_df["completo"]]
if len(incompletos) > 0:
    print("⚠️ Videos con frames sin etiquetar:")
    display(incompletos)
else:
    print("🎉 Todos los videos están completamente etiquetados.")
```
