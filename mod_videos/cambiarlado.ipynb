```python
import os
import pandas as pd
import subprocess
from pathlib import Path
```

```python
# Rutas principales
CSV_PATH = r"D:/la-u/ciclo 2025-1/Seminario/Proyecto/DATASETS_CSV/dataset_errores.csv"          # CSV con columnas: nombre_video,lado
SOURCE_DIR = Path(r"D:/la-u/ciclo 2025-1/Seminario/DATASET")    # Carpeta que contiene las subcarpetas con los videos
DEST_DIR = Path(r"D:/la-u/ciclo 2025-1/Seminario/DATASET_UNILADO")     # Carpeta destino donde se replicará la estructura
```

```python
# Cargar CSV
df = pd.read_csv(CSV_PATH)

# Crear carpeta destino si no existe
DEST_DIR.mkdir(parents=True, exist_ok=True)

for _, row in df.iterrows():
    video_name = row["nombre_video"]
    lado = row["lado"]

    # Buscar el video dentro de SOURCE_DIR recursivamente
    source_path = list(SOURCE_DIR.rglob(video_name))
    if not source_path:
        print(f"⚠️ No se encontró el video: {video_name}")
        continue

    source_path = source_path[0]

    # Construir ruta destino replicando la estructura de carpetas
    relative_path = source_path.relative_to(SOURCE_DIR)
    dest_path = DEST_DIR / relative_path
    dest_path.parent.mkdir(parents=True, exist_ok=True)

    # Si ya es "D", copiamos tal cual
    if lado == "D":
        subprocess.run([
            "ffmpeg", "-y", "-i", str(source_path),
            "-c", "copy", str(dest_path)
        ])
    # Si es "I", lo volteamos horizontalmente
    elif lado == "I":
        subprocess.run([
            "ffmpeg", "-y", "-i", str(source_path),
            "-vf", "hflip",
            "-c:a", "copy", str(dest_path)
        ])

    print(f"✅ Procesado: {video_name} → {dest_path}")
```
