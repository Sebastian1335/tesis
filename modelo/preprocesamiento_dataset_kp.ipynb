```python
import pandas as pd
```

```python
def pad_video_with_mask(df, max_frames):
    padded_dfs = []

    for video in df['video_filename'].unique():
        video_df = df[df['video_filename'] == video].copy()
        current_frames = sorted(video_df['frame'].unique())
        num_frames = len(current_frames)

        # Marcamos todos los frames originales con 1
        video_df['mask'] = 1

        if num_frames < max_frames:
            last_frame = current_frames[-1]
            last_frame_data = video_df[video_df['frame'] == last_frame]

            for new_frame in range(num_frames, max_frames):
                duplicated = last_frame_data.copy()
                duplicated['frame'] = new_frame
                duplicated['mask'] = 0  # Frame artificial
                video_df = pd.concat([video_df, duplicated])

        padded_dfs.append(video_df)

    return pd.concat(padded_dfs)
```

```python
data = pd.read_csv('D:/la-u/ciclo 2025-1/Seminario/Proyecto/DATASETS_CSV/kp_dataset_video_mode_unilado.csv')
max_frames_count = data.groupby('video_filename')['frame'].nunique().max()
padded_data = pad_video_with_mask(data, max_frames_count)
```

```python
padded_data.to_csv('keypoints_padded_with_mask_unilado.csv', index=False)
print("Archivo guardado como 'keypoints_padded_with_mask.csv'")
# print(data.groupby('video_filename')["frame"].nunique())
# print(padded_data.groupby('video_filename')["frame"].nunique())
```
